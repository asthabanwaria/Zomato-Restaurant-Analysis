# Exploratory Data Analysis (EDA)

## Business Question 1

### Which locations have the highest number of restaurants?

**Objective:**
To identify areas with the highest restaurant concentration. This helps understand market demand and competition for new restaurant businesses.

**Observation:**
BTM has the highest number of restaurants, followed by HSR and Koramangala 5th Block.

**Business Insight:**
The high concentration of restaurants in these locations indicates strong customer demand but also intense competition. Businesses planning to expand into these areas should differentiate themselves through pricing, cuisine, or customer experience.

```python
import matplotlib.pyplot as plt

top_locations = df['location'].value_counts().head(10)

plt.figure(figsize=(12,6))

top_locations.plot(kind='bar')

plt.title("Top 10 Restaurant Locations in Bangalore", fontsize=16)
plt.xlabel("Location", fontsize=12)
plt.ylabel("Number of Restaurants", fontsize=12)

plt.xticks(rotation=40)

plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()

plt.show()
```

## Interpretation

BTM has the highest concentration of restaurants, followed by HSR and Koramangala 5th Block.

This indicates that these locations have strong customer demand and a highly competitive restaurant market. Any new restaurant entering these areas should differentiate itself through pricing, cuisine, quality, or customer experience.

## Business Question 2

**Objective**

To identify areas where restaurants receive the best customer ratings.

Higher average ratings may indicate better food quality, service, or customer satisfaction.

```python
top_rating = (
    df.groupby('location')['rate']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

top_rating
```

## Business Question 2

## Which locations have the highest average restaurant ratings?

### Objective
Identify locations that consistently receive the highest customer ratings to understand where customers have the best dining experiences.

### Observation
Lavelle Road has the highest average rating (4.14), followed by Koramangala 3rd Block and St. Marks Road. Most of the top-rated locations have average ratings close to 4.0.

However, some locations such as Sankey Road have very few restaurants (27), making their average rating less reliable.

### Business Insight
Average rating alone should not be used to judge restaurant quality across locations. Locations with both high ratings and a large number of restaurants provide stronger evidence of customer satisfaction.

Koramangala 5th Block stands out because it combines:
- High average rating (4.01)
- Very large number of restaurants (2504)
- Extremely high customer engagement (over 2.2 million votes)

This indicates that Koramangala 5th Block is one of the strongest restaurant markets in Bengaluru.

```python
import matplotlib.pyplot as plt

top_rating = (
    df.groupby('location')['rate']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(12,6))

top_rating.plot(kind='bar', color='orange')

plt.title("Top 10 Locations by Average Restaurant Rating", fontsize=16)
plt.xlabel("Location")
plt.ylabel("Average Rating")

plt.xticks(rotation=40)

plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()

plt.show()
```

```python
location_analysis = (
    df.groupby('location')
      .agg(
          Average_Rating=('rate', 'mean'),
          Total_Votes=('votes', 'sum'),
          Restaurant_Count=('name', 'count')
      )
      .sort_values(by='Average_Rating', ascending=False)
)

location_analysis.head(10)
```

```python
location_summary = df.groupby('location').agg(
    Average_Rating=('rate','mean'),
    Total_Votes=('votes','sum'),
    Restaurant_Count=('name','count')
)

filtered = location_summary[location_summary['Restaurant_Count'] >= 100]

filtered.sort_values('Average_Rating', ascending=False).head(10)
```

# Business Question 3

## Which cuisines are the most popular?

### Objective

To identify the cuisines offered by the highest number of restaurants.

### Observation

North Indian cuisine is the most popular, followed by North Indian-Chinese and South Indian cuisines.

### Business Insight

The dominance of North Indian cuisine reflects strong customer demand. However, due to intense competition, new restaurants may benefit by offering unique cuisine combinations or niche food categories.

# Business Question 5

## Which restaurant types receive the highest average ratings?

### Objective

To identify which types of restaurants receive the highest customer ratings.

### Observation

Premium restaurant types such as Fine Dining, Microbrewery, Pub, and Casual Dining generally receive higher average ratings than Quick Bites and Delivery restaurants.

### Business Insight

Customers are more satisfied with restaurants that provide better dining experiences and service quality. Businesses focusing on premium dining can justify higher pricing through superior customer experience.

```python
rest_rating = df.groupby('rest_type')['rate'].mean().sort_values(ascending=False).head(10)

rest_rating
```

```python
plt.figure(figsize=(12,6))

rest_rating.plot(kind='bar', color='purple')

plt.title("Top Restaurant Types by Average Rating")
plt.xlabel("Restaurant Type")
plt.ylabel("Average Rating")

plt.xticks(rotation=45)

plt.show()
```

# Business Question 6

## Which restaurant types receive the highest customer engagement?

### Objective

To identify restaurant types that attract the highest customer interaction based on votes.

### Observation

Restaurant types offering premium dining experiences generally receive significantly higher customer votes.

### Business Insight

Higher customer engagement often indicates strong popularity and brand recognition. Restaurants with consistently high votes are more likely to attract new customers through social proof.

```python
top_votes = df.groupby('rest_type')['votes'].mean().sort_values(ascending=False).head(10)

top_votes
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

top_votes.plot(kind='bar', color='teal')

plt.title("Top Restaurant Types by Average Votes")
plt.xlabel("Restaurant Type")
plt.ylabel("Average Votes")

plt.xticks(rotation=45)

plt.show()
```

# Business Question 7

## Does higher cost lead to better restaurant ratings?

### Objective

To understand the relationship between restaurant pricing and customer ratings.

### Observation

Although many expensive restaurants have good ratings, high prices do not always guarantee higher customer satisfaction.

### Business Insight

Customer experience, food quality, and service appear to influence ratings more than price alone. Restaurants should focus on delivering value rather than simply increasing prices.

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df['approx_cost(for two people)'],
    df['rate'],
    alpha=0.3
)

plt.title("Cost vs Restaurant Rating")
plt.xlabel("Approx Cost for Two")
plt.ylabel("Rating")

plt.show()
```

# Business Question 8

## Do restaurants offering online ordering dominate the market?

### Objective

To determine how many restaurants provide online ordering services.

### Observation

A majority of restaurants offer online ordering facilities.

### Business Insight

Online ordering has become an important feature for restaurants. Businesses without online ordering may lose potential customers in today's digital food delivery market.

```python
online_counts = df['online_order'].value_counts()

online_counts
```

```python
plt.figure(figsize=(6,6))

online_counts.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.ylabel("")

plt.title("Online Order Availability")

plt.show()
```

# Business Question 9

## Does table booking influence restaurant ratings?

### Objective

To compare customer ratings between restaurants that offer table booking and those that do not.

### Observation

Restaurants providing table booking generally receive higher average ratings.

### Business Insight

Table booking enhances customer convenience and is often associated with premium dining experiences, contributing to higher customer satisfaction.

## Key Findings

1. BTM has the highest concentration of restaurants, indicating high customer demand but also intense competition.
